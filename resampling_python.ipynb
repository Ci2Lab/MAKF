# Resampling / Processing Data for GLobal Fusion

Since Julia does not have a built-in resampling funciton, load the exported intermediate results (local fusion results) and output again as CSV files to be imported for the global fusion with all estimates.

```python
# Import modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Block 4a Data

```python
# Import data from output folder
df_ts = pd.read_csv('output/local_block4a_ts.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample time series to 4h
df_ts_4h = df_ts.resample('4h').mean()
df_ts_4h = df_ts_4h.interpolate(method='linear')
```

```python
# Import satellite data from output folder
df_sat = pd.read_csv('output/local_block4a_sat.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample satellite data to 4h
df_sat_4h = df_sat.resample('4h').mean()
df_sat_4h = df_sat_4h.interpolate(method='linear')
```

```python
# Import GB-InSAR data from output folder
df_gb = pd.read_csv('output/local_block4a_gb.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample GB-InSAR data to 4h
df_gb_4h = df_gb.resample('4h').mean()
df_gb_4h = df_gb_4h.interpolate(method='linear')
```

```python
# Import Extensometer data from output folder
df_ext = pd.read_csv('output/local_block4a_ext.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample Extensometer data to 4h
df_ext_4h = df_ext.resample('4h').mean()
df_ext_4h = df_ext_4h.interpolate(method='linear')
```

```python
# Collect all data in one dataframe
df_all = pd.concat([df_ts_4h, df_sat_4h, df_gb_4h, df_ext_4h], axis=1, join='outer')
df_all.columns = ["ts_east", "ts_north", "ts_height", "sat_east", "sat_height", "gb_north", "gb_height", "ext_east", "ext_height"]
```

```python
# Export data to output folder
df_all.to_csv('output/local_all_block4a.csv')
```

```python
df_all_filtered = df_all.rolling(42, min_periods=1).mean()
```

```python
# Export data to output folder
df_all_filtered.to_csv('output/local_all_filtered_block4a.csv')
```

## Base Data

```python
# Import data from output folder
df_ts = pd.read_csv('output/local_base_ts.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample time series to 4h
df_ts_4h = df_ts.resample('4h').mean()
df_ts_4h = df_ts_4h.interpolate(method='linear')
```

```python
# Import GB-InSAR data from output folder
df_gb = pd.read_csv('output/local_base_gb.csv', parse_dates=['timestamp'], index_col='timestamp')

# Resample GB-InSAR data to 4h
df_gb_4h = df_gb.resample('4h').mean()
df_gb_4h = df_gb_4h.interpolate(method='linear')
```

```python
# Collect all data in one dataframe
df_all = pd.concat([df_ts_4h, df_gb_4h], axis=1, join='outer')
df_all.columns = ["ts_east", "ts_north", "ts_height", "gb_north", "gb_height"]
```

```python
# Export data to output folder
df_all.to_csv('output/local_all_base.csv')
```

```python
df_all_filtered = df_all.rolling(42, min_periods=1).mean()
```

```python
# Export data to output folder
df_all_filtered.to_csv('output/local_all_filtered_base.csv')
```
